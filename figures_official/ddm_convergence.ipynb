```python
import pandas as pd
from riskhistory.data import (load_ddm_rhat, load_ddm_ess)
```

```python
DATASETS = ["equalIndifference", "equalRange", "sheng2020"]
```

```python
rhat_df, ess_df = [], []
for dataset in DATASETS:
    rhat_df.append(load_ddm_rhat(dataset))
    ess_df.append(load_ddm_ess(dataset))
    
rhat_df = pd.concat(rhat_df)
ess_df = pd.concat(ess_df)
```

```python
converge_df = rhat_df.merge(ess_df, on=["model", "dataset", "variable"])
converge_df["converged"] = converge_df["rhat"] < 1.05
```

```python
for model in converge_df["model"].unique():
    print("- - - - - - - - - - - - - - - - - -")
    for dataset in ["equalIndifference", "equalRange", "sheng2020"]:
        print("=====================================")
        print(dataset)
    
        group_params_num = converge_df.query(f"dataset == '{dataset}' & model == '{model}' & ~variable.str.contains('subj') ")["variable"].nunique()
        print(model, "num parameters = ", group_params_num)
        non_converged = converge_df.query(f"dataset == '{dataset}' & model == '{model}' & ~variable.str.contains('subj') & converged == False")
        if len(non_converged["variable"].tolist()) == 0:
            print("Fully converged!")
        else:
            print(non_converged["variable"].tolist())
            print(non_converged["rhat"].tolist())
```
