```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from riskhistory.config import behavioral_data_pth
```

```python
x = pd.read_csv(behavioral_data_pth)
```

```python
y = x.query("dataset == 'equalIndifference' and sub == 13")
```

```python
hmap = y.groupby(["gain", "loss"])["accept"].mean().unstack()
hmap2 = y.groupby(["gain", "loss"])["switch"].mean().unstack()
```

```python
sns.heatmap(hmap, annot=True, cmap="coolwarm")
```

```python
sns.heatmap(hmap2, annot=True, cmap="coolwarm")
```
